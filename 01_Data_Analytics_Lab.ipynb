## အဆင့် (၁) - Database (SQL) နှင့် ချိတ်ဆက်ပြီး Data ဆွဲယူခြင်း

```python
import sqlite3
import pandas as pd
import numpy as np

# Database နှင့် ချိတ်ဆက်ခြင်း
conn = sqlite3.connect('data/hr_payroll.db')

# SQL ဖြင့် Employees Data ဆွဲထုတ်ခြင်း
query = '''
SELECT emp_id, name, gender, department, job_title, base_salary, hire_date, employment_status, performance_rating
FROM employees;
'''
df_emp = pd.read_sql_query(query, conn)
print('Data Loaded Successfully! Total Rows:', len(df_emp))
df_emp.head()
```

## အဆင့် (၂) - Python တွင် Excel Pivot Table ကဲ့သို့ အုပ်စုဖွဲ့ တွက်ချက်ခြင်း (`df.pivot_table`)
**Excel မှာ Pivot Table ဆွဲသလိုမျိုး** ဌာနအလိုက် (`Index`)၊ ကျား/မ အလိုက် (`Columns`)၊ ပျမ်းမျှလစာကို တွက်ချက်ပြခြင်း:

```python
# Excel Pivot Table in Python
pivot_salary = df_emp[df_emp['employment_status'] == 'Active'].pivot_table(
    index='department',
    columns='gender',
    values='base_salary',
    aggfunc=['count', 'mean']
)

# Format numbers nicely
pivot_salary
```

## အဆင့် (၃) - SQL + Payroll Analysis: Overtime (OT) အများဆုံး ဌာနများ စစ်ဆေးခြင်း

```python
query_ot = '''
SELECT 
    e.department,
    ROUND(SUM(p.ot_hours), 1) AS total_ot_hours,
    PRINTF('%,d MMK', SUM(p.ot_pay)) AS total_ot_cost,
    PRINTF('%,d MMK', SUM(p.gross_salary)) AS total_gross_payroll
FROM payroll_records p
JOIN employees e ON p.emp_id = e.emp_id
GROUP BY e.department
ORDER BY SUM(p.ot_hours) DESC;
'''
df_ot = pd.read_sql_query(query_ot, conn)
df_ot
```

## အဆင့် (၄) - Power BI သို့မဟုတ် Excel ထဲသို့ တိုက်ရိုက် သွင်းနိုင်ရန် Export ထုတ်ခြင်း

```python
# Cleaned & aggregated data ကို Excel/CSV အဖြစ် ထုတ်ပေးခြင်း
df_emp.to_excel('data/Cleaned_HR_Employees.xlsx', index=False)
print('✅ data/Cleaned_HR_Employees.xlsx သို့ Export ထုတ်ပြီးပါပြီ။ Power BI သို့မဟုတ် Excel တွင် ချက်ချင်း ဖွင့်နိုင်ပါပြီ!')
conn.close()
```
